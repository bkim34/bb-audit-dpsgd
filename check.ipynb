```python
from __future__ import annotations

from pathlib import Path
import numpy as np
import pandas as pd

def summarize_losses(root: Path, save_csv: bool = True) -> pd.DataFrame:
    root = root.expanduser().resolve()
    if not root.exists():
        raise FileNotFoundError(f"Root directory not found: {root}")

    rows = []
    exp_dirs = sorted([p for p in root.iterdir() if p.is_dir()])

    print(f"Scanning {root} ... found {len(exp_dirs)} experiment directories\n")

    for d in exp_dirs:
        losses_in_path  = d / "losses_in.npy"
        losses_out_path = d / "losses_out.npy"

        if not (losses_in_path.exists() and losses_out_path.exists()):
            continue

        losses_in  = np.load(losses_in_path).reshape(-1)
        losses_out = np.load(losses_out_path).reshape(-1)

        mean_in, std_in = float(losses_in.mean()), float(losses_in.std())
        mean_out, std_out = float(losses_out.mean()), float(losses_out.std())

        print(f"{d.name}")
        print(f"  in :  n={losses_in.size:>8}  mean={mean_in:.6g}  std={std_in:.6g}")
        print(f"  out:  n={losses_out.size:>8}  mean={mean_out:.6g}  std={std_out:.6g}")

        rows.append({
            "experiment": d.name,
            "path": str(d),
            "n_in": int(losses_in.size),
            "mean_in": mean_in,
            "std_in": std_in,
            "n_out": int(losses_out.size),
            "mean_out": mean_out,
            "std_out": std_out,
        })

    df = pd.DataFrame(rows).sort_values("experiment").reset_index(drop=True)

    if save_csv:
        out_path = root / "loss_summary.csv"
        df.to_csv(out_path, index=False)
        print(f"\nWrote summary CSV -> {out_path}")

    return df


if __name__ == "__main__":
    root = Path("Final_data")  # change if needed
    df = summarize_losses(root, save_csv=True)
```
